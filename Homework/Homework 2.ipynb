```python
#2
dolphincsv = pd.read_csv(r"C:\Users\bryce\GitHub\Math-300---Mathematical-Computing\Datasets\DolphinNetwork.csv")
edges = [(dolphincsv['Source'].iloc[i], dolphincsv['Target'].iloc[i]) for i in range(len(dolphincsv))]
dolphin = nx.Graph()
dolphin.add_edges_from(edges)
plt.figure(figsize = (30,30))
nx.draw(dolphin,
        node_size=600,
        with_labels=True,
        node_shape='v',
        edge_color='red',
        font_size= 30,
        font_weight='bold',
        label=True)
plt.show()
def triangle_counter(data):
    n_tri = {}
    for node in data.nodes():
        neighbors = list(data.neighbors(node))
        temp = 0
        for i in range(len(neighbors)):
            for j in range(i+1, len(neighbors)):
                if data.has_edge(neighbors[i], neighbors[j]):
                    temp +=1
        n_tri[node] = temp
    return n_tri
triangles = triangle_counter(dolphin)
plt.figure(figsize=(15,5))
plt.bar(triangles.keys(), triangles.values())
plt.title('Triangles per Dolphin')
plt.xlabel('Dolphins')
plt.ylabel('# of Triangles')
plt.xticks(rotation=90)
plt.show()

# Topless and Grin have the most with 17 each

# MN23, Quasi, Fork, SMN5, SN89, TR120, Cross, Five, Vau, Wave, TR82, Whitetip, Zig, Zipfel
# all have 0 triangles
print('The median is ')
print(sum(triangles.values())/ len(triangles))
print('Triangles')
```

```python
#4
winedf = pd.read_csv(r"C:\Users\bryce\GitHub\Math-300---Mathematical-Computing\Datasets\wine.csv")
wine = nx.Graph()

for i in winedf.index:
    wine.add_node(i, wine_class=winedf.loc[i, "Class"])


grouped_phenols = winedf.groupby("Nonflavanoid phenols").groups
for phenol_value, indices in grouped_phenols.items():
    indices = list(indices)
    for a in range(len(indices)):
        for b in range(a + 1, len(indices)):
            wine.add_edge(indices[a], indices[b])

grouped_ash = winedf.groupby("Alcalinity of ash").groups
for ash_value, indices in grouped_ash.items():
    indices = list(indices)
    for a in range(len(indices)):
        for b in range(a + 1, len(indices)):
            wine.add_edge(indices[a], indices[b])

plt.figure(figsize=(10,10))
colors = [wine.nodes[n]["wine_class"] for n in wine.nodes()]
pos = nx.spring_layout(wine, k=0.5, iterations=100)
nx.draw(wine, pos, node_color=colors, node_size=50, edge_color='gray')
plt.show()
```
