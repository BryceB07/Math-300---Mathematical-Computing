### Problem 1
Suppose $f(x)=x^2+6x+10$

1. Plot the function to visualize the minimum
2. Find the minimum using `scipy`

````python
x = np.linspace(-10, 10, 500)
y = x**2 + 6*x + 10
plt.plot(x,y)
plt.xlabel("x")
plt.ylabel("F(x)")
plt.grid()
plt.show()
#2
def objective(x):
    return (x**2 + 6*x + 10)
result = minimize(objective, x0=-10)
best_x = result.x[0]
minimum = objective(best_x)

print(best_x)
print(objective(minimum))
````


## Problem 2

Find the maximum of $50x-x^2$.

````python
def objective(x):
    return -(50*x - x**2)
result = minimize(objective, x0=-50)
print(result.x)
print(-objective(result.x[0]))
````

## Problem 3

Minimize

```python
np.sin(x)+x**2
```

using different starting values.

Do you obtain the same solution? Compute the derivative using `sympy`. When is this derivative zero?

````python
from sympy import symbols, diff, sin
def objective(x):
    return np.sin(x)+x**2
result = minimize(objective, x0=100)
print(result.x)
# I get the same solution using different starting values



x = symbols('x')
y = sin(x)+x**2
print(diff(y, x))
# derivative is 0 when x = the same x 
# value as the minimum
````

### Problem 4
Design a function with multiple minima and optimize.

````python
def objective(x):
    return 10*np.sin(x) + x**2
guess = 0
result = minimize(objective, x0=guess)
print("minima found at x=")
print(result.x)
print("and F(x) =")
print(objective(result.x))
print("guessed around")
print(guess)
````
