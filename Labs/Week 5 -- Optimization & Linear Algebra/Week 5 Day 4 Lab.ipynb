### Problem 1
Suppose 

$$\frac{dy}{dt}=1.2y.$$

instead of .5 from lecture. Solve again.

How does the graph change?



```python
def model(t,y):
    return 1.2*y
y0 = [100]
t0 = 0
t1 = 10

solution = solve_ivp(
    model,
    [t0,t1],
    y0,
    dense_output=True
)

t = np.linspace(0,10,200)
y = solution.sol(t)[0]
plt.figure(figsize=(8,5))
plt.plot(t,y)



plt.xlabel("Time")
plt.ylabel("Population")
plt.title("Exponential Growth")
plt.grid()
plt.show()
```

## Problem 4

Modify the logistic model by ranging $K$ from 0 to 50 with 100 slices.

How does the solution change?


```python
r= 0.6
K = np.linspace(1,50,100)
for i in K:
    def logistic(t,y):
        return r*y*(1-y/i)

solution = solve_ivp(
    logistic,
    [0,30],
    [5],
    dense_output=True
)

t = np.linspace(0,30,300)

plt.figure(figsize=(8,5))

plt.plot(t,
         solution.sol(t)[0])


plt.title("Logistic Growth")

plt.show()



# the solutions ceiling is where the bounds on K are
```
