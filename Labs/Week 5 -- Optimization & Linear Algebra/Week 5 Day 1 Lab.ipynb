### Problem 3
Make a 2 by 2 invertible matrix, A.
1. Compute its eigenvectors and plot them using matplotlib
2. Compute A times the eigenvectors and plot them on the same plot.
3. What do you notice?

```python
A = np.array([[2,2],[2,4]])
Ai = np.linalg.inv(A)
values, vectors = np.linalg.eig(A)

vector1 = vectors[:,0]
vector2 = vectors[:,1]

Avector1 = A@vector1
Avector2 = A@vector2

plt.plot([0,vector1[0]],[0,vector1[1]], label = 'Eigenvector 1', linewidth = 4)
plt.plot([0,vector2[0]],[0,vector2[1]], label = 'Eigenvector 2', linewidth = 4, color = 'green')

plt.plot([0,Avector1[0]],[0,Avector1[1]], label = 'A dot Eigenvector 1', linestyle='dashed', color ='yellow')
plt.plot([0,Avector2[0]],[0,Avector2[1]], label = 'A dot Eigenvector 2', linestyle='dashed', color ='red')


plt.legend()
plt.grid(True)
plt.title('Eigenvectors and A dot eigenvectors')
plt.show()
```
