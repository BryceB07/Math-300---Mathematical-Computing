### Problem 3
Create one dataset with a strong positive correlation and another with a strong negative correlation. Plot them side by side and compare their correlation coefficients.

```python
x = np.linspace(0, 50, 100)
ypos = 2 * x + (np.random.normal(0, 15, 100))
yneg = -2 * x + (np.random.normal(0, 15, 100))
plt.scatter(x, ypos, color='red')
```

```python
plt.scatter(x, yneg, color='blue')
```

```python
print(np.corrcoef(x, ypos))
print(np.corrcoef(x, yneg))
```
