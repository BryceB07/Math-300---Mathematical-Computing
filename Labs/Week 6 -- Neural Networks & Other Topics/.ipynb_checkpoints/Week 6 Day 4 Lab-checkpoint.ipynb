### Problem 1
Generate 100,000 samples using `np.random.uniform(0,1)` and compute the average so far at each sample. Plot the data. What do you notice?

```python
A = np.random.uniform(0,1,100000)
Ave = np.cumsum(A)/np.arange(1, 100001)

plt.plot(Ave, color='blue')
plt.xlabel('Number of Samples')
plt.ylabel('Value')
plt.title('Continued Average Value of Samples')
plt.grid(True)
plt.show()


# The mean aproaches 0.5 very fast
```

### Problem 2
We want to compute $\int_0^1 x^2dx$. To do so, generate random points between 0 and 1. Accept them if they are below and reject them if they are above. Compute the average of the accepted values and plot the accepted vs rejected points.

```python
n = 1000
x = np.random.uniform(0,1,n)
y = np.random.uniform(0,1,n)

inside = (x**2)>=y
plt.scatter(x[inside], y[inside], s=5, label="Accepted")
plt.scatter(x[~inside], y[~inside], s=5, label="Rejected")



plt.title("Monte Carlo on X^2 ")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')



X = np.linspace(0,1,100)
Y = X**2
plt.plot(X,Y, color='black', label='Y = X^2', linewidth=3)
plt.show()
```

### Problem 3 
Write a function that will do Problem 2 for any function and bounds.

```python
def MonteCarlo(lowerBounds, upperBounds, func, n):
    x = sp.Symbol('x')
    function = sp.lambdify(x,func, 'numpy')
    
    x = np.random.uniform(lowerBounds,upperBounds,n)
    y = np.random.uniform(lowerBounds,upperBounds,n)
    
    inside = function(x)>=y
    plt.scatter(x[inside], y[inside], s=5, label="Accepted")
    plt.scatter(x[~inside], y[~inside], s=5, label="Rejected")

    
    plt.title("Monte Carlo Approximation")
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.xlim([lowerBounds,upperBounds])
    plt.ylim([lowerBounds,upperBounds])
    plt.show()
MonteCarlo(-2,2, x**3, 1000)
```
