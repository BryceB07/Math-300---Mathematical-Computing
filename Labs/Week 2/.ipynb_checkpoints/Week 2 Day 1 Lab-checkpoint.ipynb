### Problem 2
Consider the function from Friday's quiz:
```python
def squares(maximum):

    container = []

    for i in range(maximum):
        container.append(i)

    container2 = []

    for j in container:
        container2.append(j**2)

    container3 = 0

    for k in container2:
        container3 += k

    return container3
```

What is the complexity? Justify your answer. Write a new, simpler function that does the same as `squares()`. Calculate your function's complexity and justify your answer.

The code will run 3n times so the time complexity is O(n)


````python
def squares(maximum):
    total = 0
    for i in range(maximum):
        total+= i**2
    return total
squares(5)
#time complexity would also be just O(n)
````

### Problem 5
Write a recursive function `fibonacci(n)` that computes the $n$-th Fibonacci number. Compute the time complexity of your `fibonacci(n)` function.

````python
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)
fibonacci(6)
# each time fibonacci(n) is called 2 more calls of that same function get called
#this means that the time complexity is exponential
# O(2^n)
````

### Problem 6
Write a recursive function `reverse(string)` that reverses the order of a string. Compute the time complexity of your `reverse()` function

````python
def reverse(string):
    return string[::-1]
reverse('Homework')
#O(n) because it only perorms the length of the given string
````

### Problem 7
Write a recursive function that computes the sum of integers up to input $n$.

````python
def counter(n):
    if n <= 1:
        return n
    else:
        return n + counter(n-1)
counter(10)
````
